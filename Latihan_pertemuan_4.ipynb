```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/main/penjualan.csv')
df
```

```python

plt.figure(figsize=(8, 5))

df_sorted = df.sort_values('total', ascending=False)

sns.barplot(data=df_sorted, x='nama', y='total', palette='viridis')

plt.title('Total Penjualan per Orang', fontsize=12)
plt.xlabel('Nama Karyawan')
plt.ylabel('Total Penjualan')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(5, 4))
total_per_bulan = df[['jan', 'feb', 'mar', 'apr']].sum()

plt.pie(total_per_bulan, labels=total_per_bulan.index, autopct='%1.1f%%',
        startangle=140, colors=sns.color_palette('pastel'))

plt.title('Proporsi Penjualan per Bulan', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import plotly.express as px

df_melted = df.melt(id_vars='nama', value_vars=['jan', 'mar'], var_name='Bulan',
                     value_name='Penjualan')

fig = px.line(
    df_melted, x='nama', y='Penjualan', color='Bulan', symbol='Bulan', symbol_sequence=['circle',
    'triangle-up'], color_discrete_sequence=['royalblue', 'crimson'], markers=True,

    title='Perbandingan Penjualan: Januari vs Maret',
    labels={'nama': 'Nama Karyawan', 'Penjualan': 'Jumlah Penjualan'}
)

fig.update_layout(width=800, height=500)
fig.show()
```

```python
plt.figure(figsize=(6, 5))
sns.boxplot(data=df, y='IPM', color='skyblue', width=0.4)

plt.title('Sebaran Indeks Pembangunan Manusia (IPM)', fontsize=30, fontweight='bold')
plt.ylabel('Nilai IPM (0–100)')
plt.tight_layout()
plt.show()
```

```python
g = sns.jointplot(
    data=df,
    x='PAD',
    y='KEMISKINAN',
    kind='reg',
    color='purple',
    height=6,
    scatter_kws={'alpha': 1},
    line_kws={'color': 'red'}
)

g.fig.suptitle('Analisis Mendalam: PAD vs Tingkat Kemiskinan', y=1.02, fontsize=14, fontweight='bold')
plt.show()
```
